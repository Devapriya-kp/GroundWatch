```python

keys = ["Station", "Data Acquisition Time"]

recent_counts = (
    df_recent.groupby(keys).size()
    .reset_index(name="Count")
)

historical_counts = (
    df_historical.groupby(keys).size()
    .reset_index(name="Count")
)

print("RECENT REPEATED GROUPS")

display(
    recent_counts[recent_counts["Count"] > 1]
    .sort_values("Count", ascending=False)
    .head(15)
)

print("HISTORICAL REPEATED GROUPS")

display(
    historical_counts[historical_counts["Count"] > 1]
    .sort_values("Count", ascending=False)
    .head(15)
)
```

```python

recent_repeated_keys = recent_counts[
    recent_counts["Count"] > 1
][keys]

recent_conflicts = df_recent.merge(
    recent_repeated_keys,
    on=keys,
    how="inner"
)

display(
    recent_conflicts[
        ["Station", "District", "Latitude",
         "Longitude", "Data Acquisition Time", target]
    ].sort_values(keys)
)
```

```python
# Copy both datasets
recent = df_recent.copy()
historical = df_historical.copy()

# Identify the source of each record
recent["Source"] = "Recent"
historical["Source"] = "Historical"

# Combine datasets
df_combined = pd.concat(
    [historical, recent],
    ignore_index=True
)

print("COMBINED DATASET")
print("Shape:", df_combined.shape)

df_combined.head()
```

```python
df_combined["Data Acquisition Time"] = pd.to_datetime(
    df_combined["Data Acquisition Time"],
    dayfirst=True,
    errors="coerce"
)

print(df_combined["Data Acquisition Time"].dtype)
```

```python
target = "Groundwater Level Quarterly Manual (meter)"

print("Missing readings:", df_combined[target].isnull().sum())

df_combined = df_combined.dropna(subset=[target])

print("Rows after removing missing readings:", len(df_combined))
```

```python
df_combined["Anomaly_Flag"] = (
    (df_combined[target] < 0) |
    (df_combined[target] > 100)
)

print("Suspicious readings:",
      df_combined["Anomaly_Flag"].sum())
```

```python
print("TOTAL COLUMNS:", len(df_combined.columns))

print("\nCOLUMN NAMES:")
for col in df_combined.columns:
    print(col)

print("\nDUPLICATE COLUMN NAMES:")
print(df_combined.columns[df_combined.columns.duplicated()].tolist())
```

```python
suspicious = df_combined[
    df_combined["Anomaly_Flag"] == True
]

suspicious = suspicious.sort_values(
    by=target,
    ascending=False
)

print("TOTAL SUSPICIOUS READINGS:", len(suspicious))

display(
    suspicious[[
        "Station",
        "District",
        "Latitude",
        "Longitude",
        "Data Acquisition Time",
        target,
        "Source"
    ]]
)
```

```python
negative = df_combined[df_combined[target] < 0]

print("NEGATIVE READINGS:", len(negative))

display(negative[[
    "Station",
    "District",
    target,
    "Source"
]])
```

```python
extreme = df_combined[df_combined[target] > 100]

print("READINGS ABOVE 100 METRES:", len(extreme))

display(extreme[[
    "Station",
    "District",
    "Data Acquisition Time",
    target,
    "Source"
]])
```

```python
extreme_stations = extreme["Station"].unique()

for station in extreme_stations:

    print("\nSTATION:", station)

    station_data = df_combined[
        df_combined["Station"] == station
    ].sort_values("Data Acquisition Time")

    display(station_data[[
        "Station",
        "District",
        "Data Acquisition Time",
        target,
        "Source"
    ]])
```

```python
# Create a working copy
df_final = df_combined.copy()

# Remove unnecessary leading/trailing spaces from text columns
text_cols = df_final.select_dtypes(include="object").columns

for col in text_cols:
    df_final[col] = df_final[col].str.strip()

# Standardize date format
df_final["Data Acquisition Time"] = pd.to_datetime(
    df_final["Data Acquisition Time"],
    errors="coerce"
)

# Ensure latitude and longitude are numeric
df_final["Latitude"] = pd.to_numeric(
    df_final["Latitude"], errors="coerce"
)

df_final["Longitude"] = pd.to_numeric(
    df_final["Longitude"], errors="coerce"
)

# Ensure groundwater readings are numeric
target = "Groundwater Level Quarterly Manual (meter)"

df_final[target] = pd.to_numeric(
    df_final[target], errors="coerce"
)

# Check the result
print(df_final.info())
```

```python
# Basic dataset overview

print("Dataset Shape:", df_final.shape)

print("\nSource Distribution:")
print(df_final["Source"].value_counts())

print("\nNumber of Unique Stations:")
print(df_final["Station_ID"].nunique())

print("\nNumber of Districts:")
print(df_final["District"].nunique())

print("\nDate Coverage:")
print("Start:", df_final["Data Acquisition Time"].min())
print("End:", df_final["Data Acquisition Time"].max())

print("\nGroundwater Level Statistics:")
print(df_final[target].describe())

print("\nAnomaly Flag Distribution:")
print(df_final["Anomaly_Flag"].value_counts())
```
